```python
from nltk.tokenize import word_tokenize, sent_tokenize, WordPunctTokenizer, WhitespaceTokenizer, TreebankWordTokenizer
```

- word_tokenize: Genel amaçlıdır, noktalama işaretlerini ayrı token olarak alır.

- sent_tokenize: Cümleleri ayırmak için kullanılır.

- WordPunctTokenizer: Kelimeleri ve noktalama işaretlerini ayırır. word_tokenize'a benzer ama farklı kuralları vardır.

- WhitespaceTokenizer: Sadece boşluklara bakar, noktalama işaretlerine dikkat etmez.

- TreebankWordTokenizer: Treebank corpus kurallarına göre tokenizasyon yapar. Örneğin "don't" → ["do", "n't"].

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
corpus = [
    "I love data science and machine learning.",
    "Data science is fun and exciting.",
    "Machine learning is a key part of data science."
]

# 1️⃣ CountVectorizer
count_vec = CountVectorizer()
count_matrix = count_vec.fit_transform(corpus)

print("🔢 CountVectorizer Vocabulary:")
print(count_vec.get_feature_names_out())

print("\n📊 Count Matrix:")
print(count_matrix.toarray())

# 2️⃣ TfidfVectorizer
tfidf_vec = TfidfVectorizer()
tfidf_matrix = tfidf_vec.fit_transform(corpus)

print("\n🧠 TfidfVectorizer Vocabulary:")
print(tfidf_vec.get_feature_names_out())

print("\n📈 TF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
df=pd.read_csv("yelp.csv")
```

```python
df.head()
```

```python
df["text"]=df["text"].str.lower() # Metindeki tüm kelimeler küçük harfe dönüştürülür
df["text"] = df["text"].str.replace(r'[^\w\s]', '', regex=True)  # metindeki noktalama işaretleri ve özel karakterler (harf, rakam ve boşluk dışındaki) kaldırılır
df["text"] = df["text"].str.replace(r'\d+', '', regex=True)        # metindeki tüm rakamlar  silinir
df["text"] = df["text"].str.replace(r'\n', '', regex=True)         # satır atlama (newline) karakterleri kaldırılır
df["text"] = df["text"].str.replace(r',\r', '', regex=True)        # virgül ve ardından gelen carriage return karakterleri silinir
```

```python
df.head(2)
```

```python
df["text"][0]
```

```python
df.shape
```

```python
# Sentiment Analysis
```

```python
ybw=df[(df.stars==1)|(df.stars==5)]
```

```python
ybw
```

```python
ybw.reset_index(drop=True, inplace=True)  # indeksleri sıralı hale getiriyoruz
```

```python
ybw.head(6)
```

```python
x=ybw["text"]
y=ybw["stars"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
```

```python
x.head()
```

```python
vect=CountVectorizer(stop_words="english", ngram_range=(1,2))
```

```python
x_yeni=vect.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_yeni, y, random_state=42, test_size=.2)
```

```python
nb=MultinomialNB()
```

```python
model=nb.fit(x_train, y_train)
```

```python
pred=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,pred)
```
